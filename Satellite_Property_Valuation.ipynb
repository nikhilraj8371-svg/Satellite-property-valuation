```python
import numpy as np

y_log = np.log1p(y)   # log(1 + price)
```

#Train–Validation Split

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train_log, y_val_log = train_test_split(
    X, y_log,
    test_size=0.2,
    random_state=42
)
```

#Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
```

#Training XGBoost on log(price)

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=500,
    learning_rate=0.03,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

xgb_model.fit(X_train_scaled, y_train_log)
```

```python
# Predict log(price)
y_val_pred_log = xgb_model.predict(X_val_scaled)

# Convert back to actual price scale
y_val_pred = np.expm1(y_val_pred_log)
y_val_actual = np.expm1(y_val_log)
```

#Evaluating RMSE & R²

```python
from sklearn.metrics import mean_squared_error, r2_score

rmse = np.sqrt(mean_squared_error(y_val_actual, y_val_pred))
r2 = r2_score(y_val_actual, y_val_pred)

print("XGBoost RMSE (log target):", rmse)
print("XGBoost R2 (log target):", r2)
```

#Fetching Satellite images by ESRI

```python
import math
import requests
from PIL import Image
from io import BytesIO

def deg2num(lat_deg, lon_deg, zoom):
    lat_rad = math.radians(lat_deg)
    n = 2.0 ** zoom
    xtile = int((lon_deg + 180.0) / 360.0 * n)
    ytile = int((1.0 - math.asinh(math.tan(lat_rad)) / math.pi) / 2.0 * n)
    return (xtile, ytile)

def fetch_satellite_image(lat, lon, idx, zoom=17, save_dir="data/images/train"):
    x, y = deg2num(lat, lon, zoom)

    url = (
        f"https://services.arcgisonline.com/ArcGIS/rest/services/"
        f"World_Imagery/MapServer/tile/{zoom}/{y}/{x}"
    )

    response = requests.get(url, timeout=10)
    if response.status_code == 200:
        img = Image.open(BytesIO(response.content)).convert("RGB")
        img.save(f"{save_dir}/{idx}.jpg")
```

#Robust image downloading

```python
import os
import time
from tqdm import tqdm

def download_images_from_df(df, save_dir, start_idx=0, sleep_time=0.2):
    os.makedirs(save_dir, exist_ok=True)

    for idx, row in tqdm(df.iterrows(), total=len(df)):
        img_path = f"{save_dir}/{idx}.jpg"

        # Skip if image already exists
        if os.path.exists(img_path):
            continue

        try:
            fetch_satellite_image(
                lat=row['lat'],
                lon=row['long'],
                idx=idx,
                save_dir=save_dir
            )
            time.sleep(sleep_time)

        except Exception as e:
            print(f"Failed at index {idx}: {e}")
```
